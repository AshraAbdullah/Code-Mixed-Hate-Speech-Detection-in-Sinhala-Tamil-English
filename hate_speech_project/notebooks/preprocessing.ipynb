```python
#Import libraries
import pandas as pd
import re
```

```python
#Load dataset
df = pd.read_csv("../data/raw/final_dataset.csv", encoding="utf-8")
df.head()
```

```python
#Remove empty comments
df = df.dropna(subset=["Comment"])
```

```python
#Strip whitespace
df["Comment"] = df["Comment"].astype(str).str.strip()
```

```python
#Normalize label text
df["Label"] = df["Label"].str.upper()
```

```python
#Remove URLs
df["Comment"] = df["Comment"].apply(
    lambda x: re.sub(r"http\S+|www\S+", "", x)
)
```

```python
#Remove user mentions (@name)
df["Comment"] = df["Comment"].apply(
    lambda x: re.sub(r"@\w+", "", x)
)
```

```python
#Normalize repeated characters
df["Comment"] = df["Comment"].apply(
    lambda x: re.sub(r"(.)\1{2,}", r"\1\1", x)
)
```

```python
#Lowercasing (for ML baselines)
df["Comment_clean"] = df["Comment"].str.lower()
```

```python
#Encode labels for ML
Label_map = {"NON_HATE": 0, "HATE": 1}
df["Label_id"] = df["Label"].map(Label_map)
```

```python
#Save processed data
df.to_csv("../data/processed/processed_dataset.csv", index=False, encoding="utf-8")
```
